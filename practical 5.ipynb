```python
# CBOW Model using PyTorch

# 1. Data Preparation
import torch
import torch.nn as nn

text = "I love machine learning and I love deep learning"
words = text.split()

vocab = set(words)
word_to_ix = {w: i for i, w in enumerate(vocab)}
ix_to_word = {i: w for w, i in word_to_ix.items()}
```

```python
# 2. Generate Training Data
data = []

for i in range(1, len(words) - 1):
    context = [words[i-1], words[i+1]]
    target = words[i]
    data.append((context, target))
```

```python
# 3. Define and Train CBOW Model
class CBOW(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.emb = nn.Embedding(vocab_size, 10)
        self.fc = nn.Linear(10, vocab_size)

    def forward(self, x):
        x = self.emb(x).mean(dim=0)
        return self.fc(x)

model = CBOW(len(vocab))
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

for epoch in range(100):
    for context, target in data:
        x = torch.tensor([word_to_ix[w] for w in context])
        y = torch.tensor([word_to_ix[target]])

        output = model(x)
        loss = loss_fn(output.unsqueeze(0), y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
```

```python
# 4. Output
context = ["I", "learning"]
x = torch.tensor([word_to_ix[w] for w in context])

with torch.no_grad():
    output = model(x)
    predicted = ix_to_word[output.argmax().item()]

print("Context words:", context)
print("Predicted word:", predicted)
```
